#### d. Check for duplicate columns

There are a lot of columns in this data frame. Let's see if there are any are duplicates. Note that Pandas will not allow them to have the same exact column name, so they will likely be distinct on column name but will be copies otherwise.

```python
#
# Check for duplicate columns
#-----------
# I used Google to figure this out, and adapted this example for our purposes:  
# source: https://thispointer.com/how-to-find-drop-duplicate-columns-in-a-dataframe-python-pandas/ 
def getDuplicateColumns(df):
    '''
    Get a list of duplicate columns.
    It will iterate over all the columns in dataframe and find the columns whose contents are duplicate.
    :param df: Dataframe object
    :return: List of columns whose contents are duplicates.
    '''
    duplicateColumnNames = set()
    # Iterate over all the columns in dataframe
    for x in range(df.shape[1]):
        # Select column at xth index.
        col = df.iloc[:, x]
        # Iterate over all the columns in DataFrame from (x+1)th index till end
        for y in range(x + 1, df.shape[1]):
            # Select column at yth index.
            otherCol = df.iloc[:, y]
            # Check if two columns at x 7 y index are equal
            if col.equals(otherCol):
                duplicateColumnNames.add(df.columns.values[y])
    return list(duplicateColumnNames)

duplicateColumnNames = list(getDuplicateColumns(diabetes))
print('Duplicate Columns are as follows: ')
duplicateColumnNames
```

```python
#
# drop columns from duplicates list
#-----------
# now drop list of duplicate features from our df using the .drop() method
diabetes = diabetes.drop(columns=duplicateColumnNames) 
```

### 3.2 Partition Data, Feature Selection, and Standardization

Now, we will partition our data to prepare it for the training process. Ultimately we want to use a 60% train—20% validation—20% test in this case. More data in the training set lowers bias, but then increases variance in the validation/test sets. Balancing between bias and variance with choice of these set sizes is important as we want to ensure that there is enough data to train on to get good predictions, but also want to make sure our hold-out sets are representative enough.

Work through partitioning the data into the test/train/validation sets in the chunks below. Be sure to that if you are using Ridge or LASSO, you Standardize the data. Where you do this in the workflow matters so be clear about where you are doing this and why. 

```python
#
# Partition data
#-----------

# import library
from sklearn.model_selection import train_test_split

# create y dataframe 
y = diabetes_clean['Diabetes_Number']

# create X dataframe (include everything except "Diabetes_Number", our target, 
# and "race alone or in combination with one or more other races_total population")
X = diabetes_clean.drop(['Diabetes_Number','race alone or in combination with one or more other races_total population'],
                  axis = 1)
```

Dropping CountyFIPS because we don't want the identifiers of the county to be taken into account when training our model. Geographical data will still be used to train model with the region variable

```python
#
# Feature selection
#-----------
X = X.drop(['CountyFIPS'],
                        axis = 1)
X.dtypes
```

```python
#
# Training/test split
#-----------

# set the random seed
np.random.seed(10)


# split the data so that it returns 4 values: X_train, X_test, y_train, y_test
X_train, X_temp, y_train, y_temp = train_test_split(X,                 # specify training dataset
                                                    y,                 # specify test dataset
                                                    train_size=.6,      # specify proportional split for training
                                                    test_size=.4,
                                                    random_state=10)       # specify proportional split for test
```

```python
#
# Validation split
#-----------

# validation_test_split returns 4 values: X_validation, X_test, y_train, y_test, so how do we create a 60-20-20 train-validate-test split? 
X_test, X_validate, y_test, y_validate =  train_test_split(X_temp,
                                                             y_temp,
                                                             train_size=0.5,
                                                             test_size=0.5,
                                                             random_state=10)
```

We standardize here, after the training/validation/test split, to avoid data leakage. Because standardization reflects information about the data, standardizing before splitting means that the standardization of data in the training set is influenced by the characteristics of the data in the validation/test sets and vice versa. That is, information from one split is "leaking" into another when we standardize before splitting rather than after.

```python
#
# Standardization
#-----------
# Given that we want to only standardize non-dichotomous variables, we need to find a 
# solution that will loop over only the columns we want to standardize. The code below
# identifies all non-dichotomous variables in our dataset and only standardizes those.

# load library and create instance of Standard Scaler 
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()

# identify non-dichotomous columns we want to transform
columns = list(X_test.select_dtypes(include=['number']).loc[:, X_test.nunique() > 2])

# use loop to transform training data for only columns we want to transform
for column in columns:
    X_train[column] = scaler.fit_transform(X_train[column].values.reshape(-1, 1)).flatten()

# use loop to transform validation data for only columns we want to transform
for column in columns:
    X_validate[column] = scaler.fit_transform(X_validate[column].values.reshape(-1, 1)).flatten()

# use loop to transform test data for only columns we want to transform
for column in columns:
    X_test[column] = scaler.fit_transform(X_test[column].values.reshape(-1, 1)).flatten()
```

### 4.1 Describe models

Detail the basic logic and assumptions underlying each model, its pros/cons, and why it is a plausible choice for this problem.

To predict county-level diabetes rates, we consider three models:

* Ordinary Least Squares (OLS) assumes that the relationship between our features and the diabetes rate is linear and that features are not perfectly correlated. It finds the model that minimizes the sum of squared errors without any constraint. Using OLS as a baseline, we can quantify the improvement gained from using more complex models. However, given our dataset's 80+ potential predictors, which are likely correlated, OLS is expected to be prone to overfitting and high variance

* Ridge regression addresses this multicollinearity and overfitting by adding a penalty (L2) that shrinks all coefficients towards zero but doesn't eliminates them, trading some bias for better generalization. Since we assume that many of our demographic and health factors have some influence on diabetes rates, we want to model all of them while stabilizing their collective impact

* Finally, Lasso regression (L1 penalty) helps prevent overfitting and performs feature selection by shrinking some coefficients to exactly zero. For our research question, this is very useful as it can automatically identify the most critical drivers of diabetes from our large set of potential predictors.

#### Model 1:  OLS

```python
# OLS
# Model 1 training
#-----------

from sklearn.linear_model import LinearRegression

linear_reg = LinearRegression()
linear_model = linear_reg.fit(X_train, y_train)

linear_reg_data = pd.DataFrame({
    'Feature': X.columns,
    'Coefficient': linear_model.coef_})
linear_reg_data.head(10)   
```

#### Model 2: Ridge 

```python
# Ridge.
# Model 2 training

from sklearn.model_selection import GridSearchCV
from sklearn.linear_model import Ridge
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline

# Pipeline.
ridge_pipeline = make_pipeline(StandardScaler(), Ridge())

# Test alpha values.
param_grid = {'ridge__alpha': [0.001, 0.01, 0.1, 1.0, 10, 100, 1000]}

# Grid search cross-validation.
grid_search = GridSearchCV(
    ridge_pipeline,
    param_grid,
    cv=5,
    scoring='neg_root_mean_squared_error')

# Fitting grid search and printing best alpha.
grid_search.fit(X_train, y_train)
print(f"Best alpha: {grid_search.best_params_['ridge__alpha']}")
print(f"Best CV RMSE: {-grid_search.best_score_:.4f}")

# Using best model.
best_ridge = grid_search.best_estimator_

# Coefficients.
ridge_reg_data = pd.DataFrame({
    'Feature': X.columns,
    'Coefficient': best_ridge.named_steps['ridge'].coef_})
ridge_reg_data.head(10)
```

#### Model 3:  Lasso 

```python
# Lasso 
# Model 3 training
#-----------

from sklearn.model_selection import GridSearchCV
from sklearn.linear_model import Lasso
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline

# Pipeline.
lasso_pipeline = make_pipeline(StandardScaler(), Lasso(max_iter=15000))

# Alpha values.
param_grid = {'lasso__alpha': [0.00001, 0.00005, 0.0001, 0.0005, 0.001, 0.01]}

# Cross-validation with grid search.
grid_search = GridSearchCV(
    lasso_pipeline, 
    param_grid, 
    cv=5,
    scoring='neg_root_mean_squared_error'
)

# Fitting grid search, printing best alpha.
grid_search.fit(X_train, y_train)
print(f"Best alpha: {grid_search.best_params_['lasso__alpha']}")
print(f"Best CV RMSE: {-grid_search.best_score_:.4f}")

# Best model.
best_lasso = grid_search.best_estimator_

# Showing coefficients.
lasso_reg_data = pd.DataFrame({
    'Feature': X.columns,
    'Coefficient': best_lasso.named_steps['lasso'].coef_
})
lasso_reg_data.head(10)
```

1. What is bias-variance tradeoff? Why is it relevant to machine learning problems like this one?

The bias-variance tradeoff describes the balance between a model's simplicity and its flexibility.

High bias models (like a simple linear regression) make strong assumptions and may miss important patterns, leading to underfitting.

High variance models (like very flexible ones or models overly tuned to the training data) capture noise as if it were signal, leading to overfitting.

In predicting diabetes rates across US counties, this tradeoff matters because we want our model to generalize well to new counties or future years, not just perfectly fit the data we have. Overly simple models could miss regional health or demographic trends, while overly complex ones might over tune on certain quileading to inaccurate predictions when scaled to real policy programs.

Overfitting occurs when a model learns patterns that are specific to the training data (noise) rather than general relationships that apply more broadly.
